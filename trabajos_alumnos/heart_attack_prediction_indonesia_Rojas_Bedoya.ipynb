##**Descripción del dataset seleccionado**
Disponemos de un dataset alimentado por 28 columnas y 158356 registros con antecedentes de pacientes en Indonesia. A continuación, se describe las columnas que componen el dataset.



###**Datos demográficos**
+ **Edad (int):** Edad de la persona (25-90 años)
+ **Género (str):** Género de la persona (masculino, femenino)
+ **Región (str):** zona de residencia (urbana, rural)
+ **Nivel_de_ingresos (str):** estatus socioeconómico (bajo, medio, alto)

###**Factores de riesgo clínicos**
+ **Hipertensión (bool):** presión arterial alta (1 = sí, 0 = no)
+ **Diabetes (bool):** diabetes diagnosticada (1 = sí, 0 = no)
+ **Nivel de colesterol (int):** Nivel de colesterol total (mg/dL)
+ **Obesidad (bool):** IMC > 30 (1 = Sí, 0 = No)
+ **Circunferencia de la cintura (int):** Circunferencia de la cintura en cm
+ **Historia Familiar (bool):** Antecedentes familiares de enfermedades cardíacas (1 = Sí, 0 = No)

###**Estilo de vida y factores conductuales**
+ **Fumar (str):** Hábito de fumar (Nunca, Pasado, Actual)
+ **Consumo de alcohol (str):** Consumo de alcohol (Ninguno, Moderado, Alto)
+ **Actividad física (str):** Nivel de actividad física (Bajo, Moderado, Alto)
+ **Hábitos dietéticos (str):** Calidad de la dieta (Saludable, Insalubre)

###**Factores ambientales y sociales**
+ **Exposición a la contaminación atmosférica (str):** Exposición a la contaminación (Baja, Moderada, Alta)
+ **Nivel de estrés (str):** Nivel de estrés (Bajo, Moderado, Alto)
+ **Horas de sueño (float):** Promedio de horas de sueño por noche (3-9 horas)

###**Factores de evaluación médica y del sistema de salud**
+ **Presión arterial sistólica (int):** Presión arterial sistólica (mmHg)
+ **Presión arterial diastólica (int):** Presión arterial diastólica (mmHg)
+ **Glucemia en ayunas (int):** Nivel de azúcar en sangre (mg/dL)
+ **Colesterol_hdl (int):** Nivel de colesterol HDL (mg/dL)
+ **Colesterol_ldl (int):** Nivel de colesterol LDL (mg/dL)
+ **Triglicéridos (int):** Nivel de triglicéridos (mg/dL)
+ **Resultados del electrocardiograma (str):** Resultado del electrocardiograma (normal, anormal)
+ **Enfermedad cardíaca previa (int):** Enfermedad cardíaca previa (1 = Sí, 0 = No)
+ **Uso de medicamentos (int):** Toma medicamentos para el corazón (1 = Sí, 0 = No)
+ **Participación en evaluación gratuita (int):** Participación en el programa gratuito de evaluación de salud de Indonesia (1 = Sí, 0 = No)

##**Validación adicional:**
La idea en este punto es verificar aquellos pacientes con resultados de examenes alterados en variables cómo: *blood_pressure_diastolic*, *fasting_blood_sugar*, *cholesterol_hdl*, *cholesterol_ldl, triglycerides*. Con este análisis se puede determinar la correlacion entre estos valores y el riesgo que el paciente sufra un ataque cárdiaco.



Se define los siguiente limites para determinar si un resultados en particular está dentro del rango nomal o fuera de este.

+ **blood_pressure_diastolic:** Normal < 120, elevada 120-129 mmHg, Hipertensión Grado 1 130-139, Hipertensión Grado 2 > 140 mmHg
+ **cholesterol_hdl:** En peligro < 40 mg/dL, Saludable 40-60 mg/dL, Óptimo > 60 mg/dL
+ **cholesterol_ldl:** Óptimo < 100 mg/dL, casi óptimo 100-129 mg/dL, límite alto 130-159 mg/dL, alto 160-189 mg/dL, Muy alto > 190 mg/dL
+ **triglycerides:** Saludable < 150 mg/dL, Límite alto 150-199 mg/dL, Alto 200-499 mg/dL, Muy Alto > 500 mg/dL.

```python
df['blood_pressure_diastolic'] = df['blood_pressure_diastolic'].astype(float)
df['cholesterol_hdl'] = df['cholesterol_hdl'].astype(float)
df['cholesterol_ldl'] = df['cholesterol_ldl'].astype(float)
df['triglycerides'] = df['triglycerides'].astype(float)

contador_casos = 0


for index, row in df.iterrows():
  if row['blood_pressure_diastolic'] > 120 and row['cholesterol_hdl'] < 40 and row['cholesterol_ldl'] > 130 and row['triglycerides'] > 150 and row['heart_attack'] == 1:
    contador_casos += 1

print(f"El total de casos con resultados alterados es: {contador_casos}")
```

##**Aplicación metodo del codo (Elbow) para seleccionar el número de clusters**

```python
# Metodo del Codo para encontrar el numero optimo de clusters
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 0)
    kmeans.fit(x_KM)
    wcss.append(kmeans.inertia_)
# Grafica de la suma de las distancias
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

###**Metricas: Score de Silhouette**

### **Nota aclaratoria sobre la selección de datos para el Coeficiente de Silhouette**

El cálculo del coeficiente de Silhouette, especialmente cuando se realiza para múltiples configuraciones de clusters y sobre un conjunto de datos grande, puede ser computacionalmente muy costoso. Para cada configuración de cluster, se calcula la distancia de cada punto a todos los demás puntos dentro de su propio cluster y a los puntos de los clusters vecinos.

Dado que el dataset original contiene 158,355 registros, realizar este análisis sobre el conjunto completo de datos podría requerir un tiempo de procesamiento excesivo y agotar los recursos computacionales del entorno. Para demostrar el funcionamiento del coeficiente de Silhouette y obtener una indicación de la calidad de los clusters en un tiempo razonable, se ha utilizado un subconjunto de los primeros 4,000 registros.

Es importante tener en cuenta que, si bien este subconjunto nos proporciona una visión útil, los resultados podrían variar si se ejecutara sobre el dataset completo, y la interpretación debe hacerse con esta consideración.

```python
# diccionarios vacios para ser almacenados los resultados
sse,slc = {}, {}
```

```python
from sklearn.metrics import silhouette_score,silhouette_samples
from sklearn.preprocessing import StandardScaler


for k in range(2, 11):
    # semilla de 10 para asegurar reproducibilidad
    kmeans = KMeans(n_clusters=k, max_iter=1000,random_state=10).fit(x_KM) # Algoritmo K means
    if k == 3: labels = kmeans.labels_
    clusters = kmeans.labels_ # Extraccion de labels
    sse[k] = kmeans.inertia_ # Inercia: Suma de las distancias de las muestras a su centroide mas cercano
    slc[k] = silhouette_score(x_KM,clusters) # Calculo del indice de Silhouette
```

####**Indice de Silhouette**

```python
plt.figure(figsize=(10,6))
plt.plot(list(slc.keys()), list(slc.values()))
plt.xlabel("Numero de clusters")
plt.ylabel("Score de Silhouette")
plt.show()
```

```python
import matplotlib.cm as cm

for n_clusters in range(2, 10):
    # Crear un subplot de 1 fila y 2 columnas
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)
    # El 1 subplot es para el score de silhouette
    # El rango del Ind. silhouette es [-1, 1]
    ax1.set_xlim([-0.1, 1]) # para nuestro caso lo voy a dejar en este rango
    # El (n_clusters+1)*10 es para insertar un espacio en blanco entre la silueta
     # y el grafico de la nube de puntos
    ax1.set_ylim([0, len(x_KM) + (n_clusters + 1) * 10])
    # Iniciamos el proceso de ajuste a los datos
    clusterer = KMeans(n_clusters=n_clusters,max_iter=1000, random_state=10)
    cluster_labels = clusterer.fit_predict(x_KM)
    # El score de silhouette_score da el promedio de todos los valores
    silhouette_avg = silhouette_score(x_KM, cluster_labels)
    print("Para n_clusters =", n_clusters,
          "El promedio del score de silhouette es: ", silhouette_avg)

    # Calcular el score de silhouette para cada muestra
    sample_silhouette_values = silhouette_samples(x_KM, cluster_labels)
    y_lower = 10
    for i in range(n_clusters):
        # Agregar los scores de silhouette para las muestras pertenecientes al cluster i y ordenarlos
        ith_cluster_silhouette_values =sample_silhouette_values[cluster_labels == i]
        ith_cluster_silhouette_values.sort() # Ordenar
        size_cluster_i = ith_cluster_silhouette_values.shape[0] # Tamaño de clusters
        y_upper = y_lower + size_cluster_i
        color = cm.nipy_spectral(float(i) / n_clusters) # Definir paleta de colores
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)
        # Etiquetar las graficas de silhouette con sus miembros de cluster en el medio
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
        # Calcular el nuevo y_lower para la proxima grafica
        y_lower = y_upper + 10
    ax1.set_title("Grafico de silhouette para varios clusters.")
    ax1.set_xlabel("Coeficientes de silhouette")
    ax1.set_ylabel("Label del Cluster")
    # La linea vertical para el promedio del score de silhouette de todos los valores
    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")
    ax1.set_yticks([])  # Quitar labels en y
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1]) # definir ticks para comparacion

    # 2 grafica mostrando los clusters formados
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters) # definir colores
    ax2.scatter(x_KM[:, 0], x_KM[:, 1], marker='.', s=30, lw=0, alpha=0.7,c=colors, edgecolor='k') # scatterplot
    # Etiquetar a los clusters
    centers = clusterer.cluster_centers_
    # Dibujar en color blanco los centroides de los clusters
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',c="white", alpha=1, s=200, edgecolor='k')
    # Enumerar los centroides
    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,s=50, edgecolor='k')

    ax2.set_title("Visualizacion luego de obtener clusters")
    ax2.set_xlabel("Variable X")
    ax2.set_ylabel("Variable Y")

    plt.suptitle(("Analisis de Silhouette para algoritmo de KMeans clustering "
                  "con n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()
```
